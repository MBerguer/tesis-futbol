```python
import pandas as pd
import numpy as np

def calculate_hsr_events(data, time_interval=100, threshold_hsr=2.77):
    """
    Identifica los eventos de alta aceleración (HSR) usando los datos de aceleración instantánea.

    Parámetros:
    - data: DataFrame con las posiciones de los jugadores.
    - time_interval: Intervalo de tiempo entre registros en milisegundos.
    - threshold_hsr: Umbral de aceleración para considerar un evento como HSR en m/s^2.

    Retorna:
    - hsr_data: DataFrame con los eventos HSR.
    """
    # Convertir la aceleración instantánea a un formato numérico

    # print the names of the columns
    # print(data.columns)

    # print(data[' Instantaneous Acceleration Impulse'])
    
    data[' Instantaneous Acceleration Impulse'] = pd.to_numeric(
        data[' Instantaneous Acceleration Impulse'], errors='coerce'
    )

    # Filtrar por aceleraciones HSR
    hsr_data = data[data[' Instantaneous Acceleration Impulse'] >= threshold_hsr]
    
    return hsr_data

def generate_report(data, time_interval=100, threshold_hsr=2.77):
    """
    Genera un reporte de aceleraciones HSR por jugador.

    Parámetros:
    - data: DataFrame con las posiciones de los jugadores.
    - time_interval: Intervalo de tiempo entre registros en milisegundos.
    - threshold_hsr: Umbral de aceleración para considerar un evento como HSR en m/s^2.

    Retorna:
    - report: Diccionario con el reporte por jugador.
    """
    hsr_data =  (data, time_interval, threshold_hsr)
    
    report = {}
    for player_id, group in hsr_data.groupby('Player Id'):
        report[player_id] = {
            'player_name': group[' Player Display Name'].iloc[0],
            'total_hsr_events': len(group),
            'max_acceleration': group[' Instantaneous Acceleration Impulse'].max(),
            'mean_acceleration': group[' Instantaneous Acceleration Impulse'].mean(),
            'max_speed': group['  Speed km/h'].max(),
            'mean_speed': group['  Speed km/h'].mean()
        }
    
    return report

# Cargar los datos desde el archivo CSV
data = pd.read_csv('../Session_28-10-2023.csv')


# DESCRIBE THE DATA
# print(data.describe())

# Generar el reporte con configuraciones deseadas
time_interval = 100  # intervalo de 100 ms
threshold_hsr = 2.77  # umbral de HSR en m/s^2

reporte = generate_report(data, time_interval, threshold_hsr)

# Mostrar el reporte
for player_id, metrics in reporte.items():
    print(f"Jugador {metrics['player_name']} (ID: {player_id}):")
    print(f"  Total de eventos HSR: {metrics['total_hsr_events']}")
    print(f"  Máxima aceleración: {metrics['max_acceleration']:.2f} m/s^2")
    print(f"  Aceleración media: {metrics['mean_acceleration']:.2f} m/s^2")
    print(f"  Máxima velocidad: {metrics['max_speed']:.2f} km/h")
    print(f"  Velocidad media: {metrics['mean_speed']:.2f} km/h")
    print()
```

```python
import pandas as pd
import numpy as np

def identify_wcs(data, time_interval=100, threshold_speed=5.0, threshold_acceleration=2.77, wcs_duration=10):
    """
    Identifica los escenarios de peor caso (WCS) usando los datos de velocidad y aceleración.

    Parámetros:
    - data: DataFrame con las posiciones de los jugadores.
    - time_interval: Intervalo de tiempo entre registros en milisegundos.
    - threshold_speed: Umbral de velocidad para considerar un evento como WCS en m/s.
    - threshold_acceleration: Umbral de aceleración para considerar un evento como WCS en m/s^2.
    - wcs_duration: Duración mínima en segundos para considerar un período como WCS.

    Retorna:
    - wcs_data: DataFrame con los eventos WCS.
    """
    # Convertir la velocidad de km/h a m/s
    data['  Speed (m/s)'] = data['  Speed km/h'] / 3.6
    data[' Instantaneous Acceleration Impulse'] = pd.to_numeric(data[' Instantaneous Acceleration Impulse'], errors='coerce')

    # Ordenar por jugador y tiempo
    data = data.sort_values(by=['Player Id', ' Time'])

    # Crear una columna para marcar los períodos WCS
    data['wcs'] = (data['  Speed (m/s)'] >= threshold_speed) & (data[' Instantaneous Acceleration Impulse'] >= threshold_acceleration)
    
    # Calcular la duración de cada período continuo de WCS
    data['wcs_period'] = data.groupby('Player Id')['wcs'].cumsum() - data.groupby('Player Id')['wcs'].cumcount() * data['wcs']
    
    wcs_data = pd.DataFrame()

    # print(data['wcs'])
    print(data[(data['wcs'])])
    
    for player_id, group in data.groupby('Player Id'):
        wcs_periods = group.groupby('wcs_period').filter(lambda x: len(x) * (time_interval / 1000.0) >= wcs_duration)
        wcs_data = pd.concat([wcs_data, wcs_periods])
    
    return wcs_data

def generate_wcs_report(data, time_interval=100, threshold_speed=5.0, threshold_acceleration=2.77, wcs_duration=10):
    """
    Genera un reporte de los escenarios de peor caso (WCS) por jugador.

    Parámetros:
    - data: DataFrame con las posiciones de los jugadores.
    - time_interval: Intervalo de tiempo entre registros en milisegundos.
    - threshold_speed: Umbral de velocidad para considerar un evento como WCS en m/s.
    - threshold_acceleration: Umbral de aceleración para considerar un evento como WCS en m/s^2.
    - wcs_duration: Duración mínima en segundos para considerar un período como WCS.

    Retorna:
    - report: Diccionario con el reporte por jugador.
    """
    wcs_data = identify_wcs(data, time_interval, threshold_speed, threshold_acceleration, wcs_duration)
    
    report = {}
    for player_id, group in wcs_data.groupby('Player Id'):
        # print(player_id, group)
        
        report[player_id] = {
            'player_name': player_id,
            'total_wcs_periods': group['wcs_period'].nunique(),
            'max_acceleration': group[' Instantaneous Acceleration Impulse'].max(),
            'mean_acceleration': group[' Instantaneous Acceleration Impulse'].mean(),
            'max_speed': group['  Speed (m/s)'].max(),
            'mean_speed': group['  Speed (m/s)'].mean(),
            'total_wcs_duration': group['wcs_period'].nunique() * (time_interval / 1000.0)
        }
    
    return report

# Cargar los datos desde el archivo CSV
data = pd.read_csv('../Session_28-10-2023.csv')

print(data.columns)
# print(data.describe())


# Generar el reporte con configuraciones deseadas
time_interval = 100  # intervalo de 100 ms
threshold_speed = 5.0  # umbral de velocidad en m/s
threshold_acceleration = 2.77  # umbral de aceleración en m/s^2
wcs_duration = 10  # duración mínima del WCS en segundos

reporte_wcs = generate_wcs_report(data, time_interval, threshold_speed, threshold_acceleration, wcs_duration)

# Mostrar el reporte
for player_id, metrics in reporte_wcs.items():
    print(f"Jugador {metrics['player_name']} (ID: {player_id}):")
    print(f"  Total de períodos WCS: {metrics['total_wcs_periods']}")
    print(f"  Máxima aceleración: {metrics['max_acceleration']:.2f} m/s^2")
    print(f"  Aceleración media: {metrics['mean_acceleration']:.2f} m/s^2")
    print(f"  Máxima velocidad: {metrics['max_speed']:.2f} m/s")
    print(f"  Velocidad media: {metrics['mean_speed']:.2f} m/s")
    print(f"  Duración total de WCS: {metrics['total_wcs_duration']:.2f} s")
    print()
```

```python
import pandas as pd
import numpy as np

def identify_wcs(data, time_interval=100, threshold_speed=5.0, threshold_acceleration=2.77, wcs_duration=10):
    """
    Identifica los escenarios de peor caso (WCS) usando los datos de velocidad y aceleración.

    Parámetros:
    - data: DataFrame con las posiciones de los jugadores.
    - time_interval: Intervalo de tiempo entre registros en milisegundos.
    - threshold_speed: Umbral de velocidad para considerar un evento como WCS en m/s.
    - threshold_acceleration: Umbral de aceleración para considerar un evento como WCS en m/s^2.
    - wcs_duration: Duración mínima en segundos para considerar un período como WCS.

    Retorna:
    - wcs_data: DataFrame con los eventos WCS.
    """
    # Convertir la velocidad de km/h a m/s
    data['  Speed (m/s)'] = data['  Speed km/h'] / 3.6
    
    # Asegurar que la aceleración esté en formato numérico
    data[' Instantaneous Acceleration Impulse'] = pd.to_numeric(data[' Instantaneous Acceleration Impulse'], errors='coerce')

    # Ordenar los datos por jugador y tiempo para el procesamiento secuencial
    data = data.sort_values(by=['Player Id', 'Time'])

    # Crear una columna 'wcs' para marcar los períodos que cumplen con los umbrales de velocidad y aceleración
    data['wcs'] = (data['  Speed (m/s)'] >= threshold_speed) & (data[' Instantaneous Acceleration Impulse'] >= threshold_acceleration)
    
    # Identificar los períodos continuos de WCS
    wcs_data = pd.DataFrame()
    for player_id, group in data.groupby('Player Id'):
        group['wcs_period'] = (group['wcs'] != group['wcs'].shift()).cumsum()  # Identificar cambios en el estado de WCS
        periods = group[group['wcs']]  # Filtrar solo las filas donde 'wcs' es True
        
        # Agrupar por períodos continuos
        for period_id, period_group in periods.groupby('wcs_period'):
            period_duration = len(period_group) * (time_interval / 1000.0)  # Calcular la duración del período
            if period_duration >= wcs_duration:
                wcs_data = pd.concat([wcs_data, period_group])
    
    return wcs_data

def generate_wcs_report(data, time_interval=100, threshold_speed=5.0, threshold_acceleration=2.77, wcs_duration=10):
    """
    Genera un reporte de los escenarios de peor caso (WCS) por jugador.

    Parámetros:
    - data: DataFrame con las posiciones de los jugadores.
    - time_interval: Intervalo de tiempo entre registros en milisegundos.
    - threshold_speed: Umbral de velocidad para considerar un evento como WCS en m/s.
    - threshold_acceleration: Umbral de aceleración para considerar un evento como WCS en m/s^2.
    - wcs_duration: Duración mínima en segundos para considerar un período como WCS.

    Retorna:
    - report: Diccionario con el reporte por jugador.
    """
    # Identificar los eventos WCS
    wcs_data = identify_wcs(data, time_interval, threshold_speed, threshold_acceleration, wcs_duration)
    
    report = {}
    # Agrupar por jugador para calcular estadísticas de WCS
    for player_id, group in wcs_data.groupby('Player Id'):
        # Crear un diccionario con las métricas deseadas para cada jugador
        report[player_id] = {
            'player_name': group[' Player Display Name'].iloc[0],  # Nombre del jugador
            'total_wcs_periods': len(group['wcs_period'].unique()),  # Número de períodos WCS únicos
            'max_acceleration': group[' Instantaneous Acceleration Impulse'].max(),  # Máxima aceleración registrada
            'mean_acceleration': group[' Instantaneous Acceleration Impulse'].mean(),  # Aceleración media
            'max_speed': group['  Speed (m/s)'].max(),  # Máxima velocidad registrada
            'mean_speed': group['  Speed (m/s)'].mean(),  # Velocidad media
            'total_wcs_duration': len(group) * (time_interval / 1000.0)  # Duración total de los períodos WCS en segundos
        }
    
    return report

# Cargar los datos desde el archivo CSV
data = pd.read_csv('../Session_28-10-2023.csv')

# Definir los parámetros para el análisis
time_interval = 100  # Intervalo de 100 ms entre registros
threshold_speed = 5.0  # Umbral de velocidad en m/s (equivalente a 18 km/h)
threshold_acceleration = 2.77  # Umbral de aceleración en m/s^2
wcs_duration = 10  # Duración mínima del WCS en segundos

# Generar el reporte de WCS
reporte_wcs = generate_wcs_report(data, time_interval, threshold_speed, threshold_acceleration, wcs_duration)

# Mostrar el reporte
for player_id, metrics in reporte_wcs.items():
    print(f"Jugador {metrics['player_name']} (ID: {player_id}):")
    print(f"  Total de períodos WCS: {metrics['total_wcs_periods']}")
    print(f"  Máxima aceleración: {metrics['max_acceleration']:.2f} m/s^2")
    print(f"  Aceleración media: {metrics['mean_acceleration']:.2f} m/s^2")
    print(f"  Máxima velocidad: {metrics['max_speed']:.2f} m/s")
    print(f"  Velocidad media: {metrics['mean_speed']:.2f} m/s")
    print(f"  Duración total de WCS: {metrics['total_wcs_duration']:.2f} s")
    print()
```
