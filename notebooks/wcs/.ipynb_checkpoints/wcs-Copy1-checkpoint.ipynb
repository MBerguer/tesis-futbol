```python
import pandas as pd
import numpy as np

def calcular_distancia(intervalo):
    # print("[calcular_distancia]")
    # Calcula la distancia total en el intervalo
    # Usar la fórmula de la distancia entre coordenadas (lat, lon)
    distancia_total = 0.0
    for i in range(1, len(intervalo)):
        lat1, lon1 = intervalo.iloc[i-1][[' Lat', ' Lon']]
        lat2, lon2 = intervalo.iloc[i][[' Lat', ' Lon']]
        distancia_total += haversine_distance(lat1, lon1, lat2, lon2)
    return distancia_total

def calcular_hsr(intervalo):
    # print("[calcular_hsr]")
    # Calcula la distancia en alta velocidad (HSR > 19.8 km/h)
    hsr_total = intervalo[intervalo['  Speed km/h'] > 19.8]['  Speed km/h'].sum()
    return hsr_total

def calcular_sprints(intervalo):
    # print("[calcular_sprints]")
    # Calcula la distancia en sprints (Velocidad > 25.2 km/h)
    sprint_total = intervalo[intervalo['  Speed km/h'] > 25.2]['  Speed km/h'].sum()
    return sprint_total

def contar_aceleraciones(intervalo):
    # print("[contar_aceleraciones]")
    # Cuenta las aceleraciones significativas (> 3 m/s^2)
    aceleraciones = np.sum(intervalo[' Instantaneous Acceleration Impulse'] > 3)
    return aceleraciones

def contar_desaceleraciones(intervalo):
    # print("[contar_desaceleraciones]")
    # Cuenta las desaceleraciones significativas (< -3 m/s^2)
    desaceleraciones = np.sum(intervalo[' Instantaneous Acceleration Impulse'] < -3)
    return desaceleraciones

def calcular_hml_distance(intervalo):
    # print("[calcular_hml_distance]")
    # Calcula la distancia en alta carga mecánica (HMLD)
    # Suponiendo que HMLD es para velocidades > 19.8 km/h y aceleraciones > 3 m/s^2
    hml_distance = intervalo[(intervalo['  Speed km/h'] > 19.8) & (intervalo[' Instantaneous Acceleration Impulse'] > 3)]['  Speed km/h'].sum()
    return hml_distance

def calcular_amp(intervalo):
    # print("[calcular_amp]")
    # Calcula la potencia metabólica promedio (AMP)
    # Esto es una simplificación para fines de ejemplo
    amp = intervalo['  Speed km/h'].mean()  # En lugar de potencia metabólica exacta
    return amp

def haversine_distance(lat1, lon1, lat2, lon2):
    # print("[haversine_distance]", lat1, lon1, lat2, lon2)
    # Calcula la distancia entre dos puntos en la Tierra utilizando la fórmula de Haversine
    R = 6371e3  # Radio de la Tierra en metros
    phi1 = np.radians(lat1)
    phi2 = np.radians(lat2)
    delta_phi = np.radians(lat2 - lat1)
    delta_lambda = np.radians(lon2 - lon1)
    
    a = np.sin(delta_phi/2) * np.sin(delta_phi/2) + \
        np.cos(phi1) * np.cos(phi2) * \
        np.sin(delta_lambda/2) * np.sin(delta_lambda/2)
    
    c = 2 * np.arctan2(np.sqrt(a), np.sqrt(1-a))
    
    distancia = R * c  # en metros
    return distancia
```

```python


size_segundo = 10

def detectar_peores_escenarios(datos_csv, ventanas_de_tiempo):
    print("[detectar_peores_escenarios]", datos_csv, ventanas_de_tiempo)
    peores_escenarios = {}

    jugadores = datos_csv['Player Id'].unique()
    
    for jugador in jugadores:
        print("jugador:", jugador)
        datos_jugador = datos_csv[datos_csv['Player Id'] == jugador]


        peor_escenario_jugador = {}
        
        for ventana in ventanas_de_tiempo:
            size_ventana = ventana*size_segundo
            ultima_ventana_jugador = len(datos_jugador) - size_ventana
            # print("datos_jugador", datos_jugador)
            print("ultima_ventana_jugador", ultima_ventana_jugador)
            
            for i in range(1, ultima_ventana_jugador, size_ventana):
                j = i + size_ventana
                
                
                intervalo = datos_jugador.iloc[i:j]
                
                distancia_total = calcular_distancia(intervalo)
                hsr_total = calcular_hsr(intervalo)
                sprint_total = calcular_sprints(intervalo)
                aceleraciones = contar_aceleraciones(intervalo)
                desaceleraciones = contar_desaceleraciones(intervalo)
                hml_distance = calcular_hml_distance(intervalo)
                amp = calcular_amp(intervalo)

                # print(i, j, distancia_total, hsr_total)
                
                # distancia_total
                if (not ("distancia_total" in peor_escenario_jugador)) or ("distancia_total" in peor_escenario_jugador and distancia_total > peor_escenario_jugador["distancia_total"]["distancia_total"]):
                    peor_escenario_jugador["distancia_total"] = {
                        "i_ventana": i,
                        "tamaño_ventana": ventana,
                        "distancia_total": distancia_total,
                        "hsr_total": hsr_total,
                        "sprint_total": sprint_total,
                        "aceleraciones": aceleraciones,
                        "desaceleraciones": desaceleraciones,
                        "hml_distance": hml_distance,
                        "amp": amp,
                        "inicio_ventana": intervalo.iloc[0][" Time"],
                        "fin_ventana": intervalo.iloc[-1][" Time"]
                    }
                
                # hsr_total
                if (not ("hsr_total" in peor_escenario_jugador)) or ("hsr_total" in peor_escenario_jugador and hsr_total > peor_escenario_jugador["hsr_total"]["hsr_total"]):
                    peor_escenario_jugador["hsr_total"] = {
                        "i_ventana": i,
                        "tamaño_ventana": ventana,
                        "distancia_total": distancia_total,
                        "hsr_total": hsr_total,
                        "sprint_total": sprint_total,
                        "aceleraciones": aceleraciones,
                        "desaceleraciones": desaceleraciones,
                        "hml_distance": hml_distance,
                        "amp": amp,
                        "inicio_ventana": intervalo.iloc[0][" Time"],
                        "fin_ventana": intervalo.iloc[-1][" Time"]
                    }
                
                # sprint_total
                if (not ("sprint_total" in peor_escenario_jugador)) or ("sprint_total" in peor_escenario_jugador and sprint_total > peor_escenario_jugador["sprint_total"]["sprint_total"]):
                    peor_escenario_jugador["sprint_total"] = {
                        "i_ventana": i,
                        "tamaño_ventana": ventana,
                        "distancia_total": distancia_total,
                        "hsr_total": hsr_total,
                        "sprint_total": sprint_total,
                        "aceleraciones": aceleraciones,
                        "desaceleraciones": desaceleraciones,
                        "hml_distance": hml_distance,
                        "amp": amp,
                        "inicio_ventana": intervalo.iloc[0][" Time"],
                        "fin_ventana": intervalo.iloc[-1][" Time"]
                    }
                
                # aceleraciones
                if (not ("aceleraciones" in peor_escenario_jugador)) or ("aceleraciones" in peor_escenario_jugador and aceleraciones > peor_escenario_jugador["aceleraciones"]["aceleraciones"]):
                    peor_escenario_jugador["aceleraciones"] = {
                        "i_ventana": i,
                        "tamaño_ventana": ventana,
                        "distancia_total": distancia_total,
                        "hsr_total": hsr_total,
                        "sprint_total": sprint_total,
                        "aceleraciones": aceleraciones,
                        "desaceleraciones": desaceleraciones,
                        "hml_distance": hml_distance,
                        "amp": amp,
                        "inicio_ventana": intervalo.iloc[0][" Time"],
                        "fin_ventana": intervalo.iloc[-1][" Time"]
                    }
                    
                # desaceleraciones
                if (not ("desaceleraciones" in peor_escenario_jugador)) or ("desaceleraciones" in peor_escenario_jugador and desaceleraciones > peor_escenario_jugador["desaceleraciones"]["desaceleraciones"]):
                    peor_escenario_jugador["desaceleraciones"] = {
                        "i_ventana": i,
                        "tamaño_ventana": ventana,
                        "distancia_total": distancia_total,
                        "hsr_total": hsr_total,
                        "sprint_total": sprint_total,
                        "aceleraciones": aceleraciones,
                        "desaceleraciones": desaceleraciones,
                        "hml_distance": hml_distance,
                        "amp": amp,
                        "inicio_ventana": intervalo.iloc[0][" Time"],
                        "fin_ventana": intervalo.iloc[-1][" Time"]
                    }
                    
                # hml_distance
                if (not ("hml_distance" in peor_escenario_jugador)) or ("hml_distance" in peor_escenario_jugador and hml_distance > peor_escenario_jugador["hml_distance"]["hml_distance"]):
                    peor_escenario_jugador["hml_distance"] = {
                        "i_ventana": i,
                        "tamaño_ventana": ventana,
                        "distancia_total": distancia_total,
                        "hsr_total": hsr_total,
                        "sprint_total": sprint_total,
                        "aceleraciones": aceleraciones,
                        "desaceleraciones": desaceleraciones,
                        "hml_distance": hml_distance,
                        "amp": amp,
                        "inicio_ventana": intervalo.iloc[0][" Time"],
                        "fin_ventana": intervalo.iloc[-1][" Time"]
                    }
                    
                # amp
                if (not ("amp" in peor_escenario_jugador)) or ("amp" in peor_escenario_jugador and amp > peor_escenario_jugador["amp"]["amp"]):
                    peor_escenario_jugador["amp"] = {
                        "i_ventana": i,
                        "tamaño_ventana": ventana,
                        "distancia_total": distancia_total,
                        "hsr_total": hsr_total,
                        "sprint_total": sprint_total,
                        "aceleraciones": aceleraciones,
                        "desaceleraciones": desaceleraciones,
                        "hml_distance": hml_distance,
                        "amp": amp,
                        "inicio_ventana": intervalo.iloc[0][" Time"],
                        "fin_ventana": intervalo.iloc[-1][" Time"]
                }

        print(jugador)
        print(peor_escenario_jugador)
        peores_escenarios[jugador] = peor_escenario_jugador
    
    return peores_escenarios

print("Cargando datos...")
datos_csv = pd.read_csv('../Session_28-10-2023.csv')

# Definir las ventanas de tiempo en segundos
ventanas_de_tiempo = [10]  # 1 minuto, 5 minutos, 10 minutos

# Detectar peores escenarios
print("Detectando peores escenarios...")
peores_escenarios = detectar_peores_escenarios(datos_csv, ventanas_de_tiempo)

# Mostrar resultados
print("Peores escenarios detectados:")
print(peores_escenarios)
```
